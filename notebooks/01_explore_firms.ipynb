# Explore FIRMS labels
Exploration only. Run `make download-labels` first. Pipeline logic lives in `src/wildfire`.

```python
import geopandas as gpd, pandas as pd, matplotlib.pyplot as plt
raw = gpd.read_parquet('../data/raw/firms/firms_all.parquet')
flt = gpd.read_parquet('../data/raw/firms/firms_filtered.parquet')
print(len(raw), len(flt))
raw.confidence.value_counts()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
raw.groupby('acq_date').size().plot(ax=ax[0], title='Detections per day (all)')
flt.plot(ax=ax[1], markersize=1, column='frp', legend=True); ax[1].set_title('Filtered detections coloured by FRP')
```
